```python
pinn='1234'
amountn=60000

for i in range(3):
    pin=input("Enter the pin number: ")
    if pin==pinn:
        amount=int(input("Enter amount: "))
        if amount<= amountn:
            if amount%1000==0:
                print("Amount withdraw.")
                print(f'Withdrawn amount={amount}.')
                print(f'Remaining Amount={amountn-amount}.')
                break                                          # break prevents the loop to run again and again 
            else:
                print("Amount should be multiple of 1000.")
        else:
            print("Insufficent amount.")
    else:
        print('Incorrect pin.')
else:
    print('Account Blocked')
```

```python
# use shlould be able to make at least 5 orders
# user should be able to stop even at 2 order -----basically the loop should stop 
# apply match case within match case 

print('1.Breakfast')
print('2.Dessert')
print('3.Dinner')
print('4.Launch')
print('5.Snacks')
print('6. Drinks')

for choice in range(5):
    
    choice= input('Enter your choice: ')
    if choice=='0':
        print('Order completed')
        break
        
    match choice:
    
        case '1':
            print('Your have ordered breakfast')
            print('1.Tea')
            print('2.Coffee')
            print('3.Milk')
            b_choice=input('Enter breakfast choice: ')
            match b_choice:
                case '1':
                    print('You get Tea')
                case '2':
                    print('You get coffee')
                case '3':
                    print('You get Milk')
                case _:
                    print('Invalid order')
        case '2':
            print('You have ordered Dessert.')
            print('1.Icecream')
            print('2.Sweet')
            d_choice=input('Enter dessert choice: ')
            match d_choice:
                case '1':
                    print('You get Icecream')
                case '2':
                    print('You get Sweet')
                case _:
                    print('Invalid order')
        case '3':
            print('You hvae ordered Dinner.')
        case '4':
            print('You hvae ordered Launch.')
        case '5':
            print('You hvae ordered Snacks.')
        case '6':
            print('You hvae ordered Drinks.')
            
        case _:
            print('You have no order.')
```

Write a program that asks the user to guess a number between 1 and 10. The secret number is 7 (hardcoded). Give the user up to 5 attempts using a for loop. If they guess correctly, print "Correct!" and stop asking (even if attempts remain). If they run out of attempts without guessing, print "Out of attempts! The number was 7."

```python
for num in range(5):
    num=int(input('Enter number: '))
    if num==7:
        print('Correct')
        break
    else:
        print('incorrect')
else:
    print('Out of attempts! The number was 7')
        
```

Write a program using match-case that acts as a simple calculator. Ask the user to enter an operator (+, -, *, /) and two numbers. Use match to perform the correct operation and print the result. If the operator doesn't match any case, print "Invalid operator".

```python
a=int(input('enter first number: '))
b=int(input('enter second number: '))

print('Operations to be performed:')
print('1.Addition')
print('2.Subtraction')
print('3.Division')
print('4.Multiplication')

op=input('Enter operation:')
match op:
    case '1':
        result=a+b
        print(f'Result: {result}')
    case '2':
        result=a-b
        print(f'Result: {result}')
    case '3':
        if b==0:
            print('Cannot perform division by Zero')
        else:
            result= a/b
            print(f'Result:{result}')
    case '4':
        result= a*b
        print(f'Result: {result}')
    case _:
        print('Invalid Operator')
        
```

Write a program that asks the user to enter numbers one at a time (up to 5 times) using a for loop. If the user enters a negative number, stop the loop immediately using break and print "Negative number found, stopping." If the loop finishes all 5 entries without ever finding a negative number, print "All 5 numbers were positive." (using for-else).

```python
for num in range(5):
    
        num=int(input('Enter a number'))
        
        if num<0:
                print('Negative number found,stopping')
                break
else:
        print('All 5 numbers were positive')

        
```

Write a program that takes a word from the user and counts how many vowels (a, e, i, o, u) it contains. Use a for loop to go through each character in the string one by one, and an if to check if it's a vowel. Print the total count at the end.

```python
word=input('enter a word')
count=0

for letter in word:
    if letter in 'aeiou':
        count=count+1
print(f'Numnber of counts:{count}')
    
   
```

```python
secret_num=7

while True:
    guess_num=int(input('Enter a guess number'))

    if guess_num==secret_num:
        print("correct guess")
        break
    else:
        print("Incoreect guess")
        
 
                  
```

```python
secret_num=7
guess_num=0

while guess_num!=secret_num:
    print(f"Previous number:{guess_num}")
    guess_num=int(input('Enter a guess number'))

    if guess_num==secret_num:
        print("correct guess")
    else:
        print("Incoreect guess")
```

```python
## secret_num=7
guess_num=0
no_attempts=6
while guess_num!=secret_num:
    print(f"Previous number:{guess_num}")
    guess_num=int(input('Enter a guess number:'))

    if guess_num==secret_num:
        print("correct guess")
    else:
        print("Incoreect guess")
        no_attempts=no_attempts-1
        print(f"Remaining attempts:{no_attempts}")
        print("\n")
        
    if no_attempts==0:
        print("No more attepts")
        break
    
```
