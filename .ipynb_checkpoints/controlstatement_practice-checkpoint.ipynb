An electricity company charges customers based on the following rules:

The first 100 units are charged at Rs. 5 per unit.
The next 100 units (101–200) are charged at Rs. 7 per unit.
Any units above 200 are charged at Rs. 10 per unit.

Write a program that accepts the number of units consumed and prints the total electricity bill.

```python
unit=int(input('Enter units:'))

if unit<=100 and unit>0:
         bill=unit*5
         print(f'Total Electricity Bill:Rs{bill}')
elif unit>=101 and unit<=200:
          bill=unit*7
          print(f'Total Electricity Bill:Rs{bill}')
elif unit>200:
          bill=unit*10
          print(f'Total Electricity Bill: Rs{bill}')
```

A company gives bonuses according to these rules:

Employees with 10 or more years of service receive a 20% bonus.
Employees with 5 to 9 years of service receive a 10% bonus.
Employees with less than 5 years of service receive no bonus.

The program should ask for the employee's salary and years of service, then display the bonus amount and the final salary after adding the bonus.


```python
exp_year=int(input('Enter year of experiemce:'))
salary=int(input('Enter salary: '))
if exp_year>=10:
    bonus=0.2*salary
    salary=bonus+salary
    print(f'Final Salary: {salary}')
elif exp_year>=5 and exp_year<=9:
    bonus=0.1*salary
    salary=bonus+ salary
    print(f'Final salary: {salary}')
elif exp_year<5:
    print(f'Final salary: {salary}')
```

The user enters the lengths of three sides.

Your program should determine whether those three sides can form a valid triangle. If they do, print "Valid Triangle"; otherwise, print "Invalid Triangle."

```python
a=int(input('Enter one side:'))
b=int(input('Enter one side:'))
c=int(input('Enter one side:'))

if a+b>c and b+c>a and c+a>b:
       print('Valid triangle.')
else:
       print('Invalid triangle.')
```

An online store offers discounts based on the total purchase amount.

Rs. 5,000 or more → 5% discount
Rs. 10,000 or more → 10% discount
Rs. 20,000 or more → 20% discount

Display:

Original amount
Discount amount
Final payable amount

```python
p_amount=int(input('Enter purchased amount:'))

if p_amount>=20000:
    d_amount=0.2*p_amount
    f_amount=p_amount-d_amount
    print(f'Original Amount: {p_amount}')
    print(f'Discount Amount: {d_amount}')
    print(f'Final Amount: {f_amount}')
elif p_amount>=10000:
    d_amount=0.1*p_amount
    f_amount=p_amount-d_amount
    print(f'Original Amount: {p_amount}')
    print(f'Discount Amount: {d_amount}')
    print(f'Final Amount: {f_amount}')
elif p_amount>=5000:
    d_amount=0.05*p_amount
    f_amount=p_amount-d_amount
    print(f'Original Amount: {p_amount}')
    print(f'Discount Amount: {d_amount}')
    print(f'Final Amount: {f_amount}')
else:
    print('Discount not offered')
    print(f'Original Amount: {p_amount}')

    
```

The ticket price depends on age.

Below 5 years → Free
5–12 years → Rs. 150
13–59 years → Rs. 300
60 years and above → Rs. 200

Ask the user for their age and display the ticket price.

```python
age=int(input('Enter age: '))

if age<5 and age>=0:
    print('Free ticket')
elif age<=12 and age>=5:
    t_price=150
    print(f'Ticket:Rs{t_price}')
elif age<=59 and age>=13:
    t_price=300
    print(f'Ticket:Rs{t_price}')
else:
    t_price=200
    print(f'Ticket:Rs{t_price}')
```
