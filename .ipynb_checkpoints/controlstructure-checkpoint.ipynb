```python
# WAP to find a number is positive or negative
number=int(input('Enter a number: '))

if number>=0:
    print(f'The number is positive.')
else:
    print(f'The number is negative.')
```

```python
# elif statement 
# elif can be added as many as you can betwenn if else
# elif has condition


temp=float(input('enter temperatute: ')) # input 13

if temp>=20: # 13<20
    print('IT is hot toaday.')
elif temp>=10 and temp<20: # 13>10( true) and 13<20( true) ((TRUE))
    print("it is warm today.")
else:
    print('IT is cold today.')


    
```

```python
#wap to check a number is odd or even

num=int(input('enter a number: '))  # input =3

if num%2==0: # 3%2 has a remainder........so is odd number
    print('the number is even')
else:
    print('the number is odd')
```

```python
#wap to get student grade 100-80( distiction) 79-70(first division) 69-60(second)........ less then 40 is fail

student_grade=int(input('Enter grade of student:'))

if student_grade<=100 and student_grade>=80:
    print('distiction')
elif student_grade<=79 and student_grade>=70:
    print('Grade A')
elif student_grade<=69 and student_grade>=60:
    print('Grade B')
elif student_grade<=59 and student_grade>=50:
    print('Grade C')
elif student_grade<=49 and student_grade>=40:
    print('Grade D')
else:
    print('fail')

```
