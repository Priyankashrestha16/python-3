```python
class Employee:
    def __init__(self,emp_id,emp_name,emp_type,emp_position):
        self.emp_id=emp_id        #storing local variable in instance variable
        self.emp_name=emp_name
        self.emp_type=emp_type
        self.emp_position=emp_position

    def employee_detail(self):        #here the self represents particular object like....e1.employee_detail() here self represents e1 object
                                       #e2.employee_detail() here self would represent e2 object and access e2 data in the function call
        print(self.emp_id)     #we must use instance variable because local variable cannot be called outside constructor
        print(self.emp_name)
        print(self.emp_type)
        print(self.emp_position)


        
```

```python
class customer:
    def __init__(self,customer_id,customer_name,customer_address,customer_type,customer_contact):
        self.customer_id=customer_id
        self.customer_name=customer_name
        self.customer_address=customer_address
        self.customer_type=customer_type
        self.customer_contact=customer_contact

    def customer_details(self):
        print(f'{self.customer_id}')
        print(f'{self.customer_name}')
        print(f'{self.customer_address}')
        print(f'{self.customer_type}')
        print(f'{self.customer_contact}')
```

```python
class order(customer):           #here order being the child class and customer being the child class   (order depends on customer) 
    def __init__(self,order_id,order_type,order_destination,customer_id,customer_name,customer_address,customer_type,customer_contact):    #parent class constructor arguments needs to be passed here as well
        super().__init__(customer_id,customer_name,customer_address,customer_type,customer_contact)    #only customer arguments from constructorSS are passed not (self)
        self.order_id=order_id
        self.order_type=order_type
        self.order_destination=order_destination

    def order_details(self):
        print(f'{self.order_id}')
        print(f'{self.order_type}')
        print(f'{self.order_destination}')
```

```python
o1=order(100,'Veg','downhall',103,'ronn','chilly','vegan','93227367263')
o1.order_details()
o1.customer_details()
```

Encapsulation

```python
''' 
Data hiding

Access Modifiers
1.Public  --> self.var_name

2. private  --> self.__var_name => (__) double underscore

3. protected --> self._var_name => (_) single underscore

'''
```

```python
''' 

Rule:

variable cannot be accessed by its own class 

'''
```

```python
class customer:
    def __init__(self,customer_id,customer_name,customer_address,customer_type,customer_contact):

        # Using private access modifier
        self.__customer_id=customer_id
        self.__customer_name=customer_name
        self.__customer_address=customer_address
        self.__customer_type=customer_type
        self.__customer_contact=customer_contact

    def customer_details(self):
        print(f'{self.__customer_id}')
        print(f'{self.__customer_name}')
        print(f'{self.__customer_address}')
        print(f'{self.__customer_type}')
        print(f'{self.__customer_contact}')
```

```python
class order(customer):           
    def __init__(self,order_id,order_type,order_destination,customer_id,customer_name,customer_address,customer_type,customer_contact):    
        super().__init__(customer_id,customer_name,customer_address,customer_type,customer_contact)    
        self.order_id=order_id              # public access modifier 
        self.order_type=order_type
        self.order_destination=order_destination

    def order_details(self):
        print(f'{self.order_id}')
        print(f'{self.order_type}')
        print(f'{self.order_destination}')
```

```python
o=order(11,34,'jhapa','23','rita','hdshs','role','23424244224')
o.customer_details()
print()
o.order_details()
```

```python
c=customer('23','rita','hdshs','role','23424244224')
c.customer_details()
```

```python
o.customer_id        # after using private access modifier the variable can only be used within the class only

c.customer_id         # private variable cannot be accessed by its own class object as well
```

Protected --> can be accssed using (_) before accessing variable name

```python
class customer:
    def __init__(self,customer_id,customer_name,customer_address,customer_type,customer_contact):

        # Using protected access modifier
        self._customer_id=customer_id
        self._customer_name=customer_name
        self._customer_address=customer_address
        self._customer_type=customer_type
        self._customer_contact=customer_contact

    def customer_details(self):
        print(f'{self._customer_id}')
        print(f'{self._customer_name}')
        print(f'{self._customer_address}')
        print(f'{self._customer_type}')
        print(f'{self._customer_contact}')
```

```python
class order(customer):            
    def __init__(self,order_id,order_type,order_destination,customer_id,customer_name,customer_address,customer_type,customer_contact):    
        super().__init__(customer_id,customer_name,customer_address,customer_type,customer_contact)   
        self.order_id=order_id
        self.order_type=order_type
        self.order_destination=order_destination

    def order_details(self):
        print(f'{self.order_id}')
        print(f'{self.order_type}')
        print(f'{self.order_destination}')
```

```python
class Khalti:  
    balance=4000                      
    def load_wallet(self,amount):                 #all load_wallete() function does the same work.
        print(f'{amount} is deposited in khalti wallet.')
        Khalti.balance+amount
        print(f'Available Balance= {Khalti.balance}')

    def pay(self,amount):
        print(f'{amount} is deposited in khalti wallet.')
        Khalti.balance-=amount
        print(f'Available Balance after withdrawl= {self.balance}')

class Esewa:
    balance=40
    def load_wallet(self,amount):
        print(f'{amount} is deposited in esewa wallet.')
        Esewa.balance+=amount
        print(f'Available Balance= {Esewa.balance}')

class Paywell:
    balance=400
    def load_wallet(self,amount):
        print(f'{amount} is deposited in paywell wallet.')
        Paywell.balance+=amount

        print(f'Available Balance= {Paywell.balance}')
```

```python
def load_function(balance,obj):       #creating a new fucntion that can perform all class fucntion using specific object
    obj.load_wallet(balance)
    return obj.balance           # fucntion is called with the argumenrs needed fot the function

def pay_function(balance,obj):
    obj.pay(balance)
    return obj.balance


load_function(3500,Esewa())         # the new fucntion helps to execute sepecific class 

print()
load_function(3500,Khalti()) 
pay_function(300,Khalti())  
```

```python
''' 
1.we need to create an abtract class for abstraction
2.if we create abstract class then the class must have at least one abastract method

'''
```

```python
#abstract base class
from abc import ABC, abstractmethod 
```

```python
class MobileBanking(ABC):
    @abstractmethod
    def method(self,account_no,balance):
        pass

```

```python
class Account(MobileBanking):
    def amount(self):
        print('amount')

    def method(self,account_no, balance):
        print(f'{self.account_no} has balance : {self.balance}')
```

```python
a=Account()

a.amount()
a.method(113331344,8000)

```
