All arguments and all keyword arguments

```python
'''
all arguments ---> *args (tuple)
all keyword arguments ---> **kwargs(dictonary)

'''
```

```python
def emp_details(*args):          #args name is not compulsory it can be any other name as well
    print(args)
```

```python
emp_details('john','east',35)      # can pass as many argumets possible with args
```

```python
def emp_detail(address,*args):   #if any arguments are compulsory then it has to be added before *args
    print(address)
    print(args)
```

```python
emp_detail('lalitpur','john','east',45)    
```

```python
def kw_emp_details(**kwargs):
    print(kwargs)
```

```python
kw_emp_details(emp_name='john',emzone='east',emage=45)
```

```python
# another simple function

def odd_even(val):
    if val%2==0:
        print('Even')
    else:
        print('Odd')

odd_even(56)
```

```python
#lambda function

od_even= lambda val: print(f'{val}: even number') if val%2==0 else print(f'{val}:odd number')
```

```python
od_even(3)
```

```python
#another simple function

def add_data():
    data=[]
    for i in range(1,11):
        data.append(i)
    print(data)

add_data()
```

```python
# using lambda function

ad_data=lambda : [i for i in range(1,11)]

ad_data()
```

```python
# if any function has for and if together at once then it is compulsory to have both inside a list []

ad_data=lambda : [f'{i}: even number' if i%2==0 else f'{i}:odd number' for i in range(1,11)]

ad_data()
```

```python
#simple task

prices=[100,250,75,400,120,310]
cheap_items=[]
expensive_items=[]

for price in prices:
    if price>=200:
        expensive_items.append(price)
        expensive_items.sort()
    else:
        cheap_items.append(price)
        cheap_items.sort()

print(f'Cheap Items:{cheap_items}')
print(f'Expensive Items:{expensive_items}')
```

```python
c_it=lambda : [price for price in prices if price<=200]
e_it= lambda :[price for price in prices if price>200]

print(c_it())
print(e_it())
```

```python
prices=[100,250,75,400,120,310]
cheap_item=[]
expensive_item=[]

app=lambda : [f'{expensive_item.append(i)}' if i>=200 else f'{cheap_item.append(i)}' for i in prices]

app()
```
