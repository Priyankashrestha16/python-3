```python
'''
Q8. Leap Year Checker
Write a program to determine if a given year variable is a leap year. A year is a leap year if it is 
divisible by 4. However, century years (ending in 00) must also be divisible by 400 to be a leap 
year. Use nested if structures to solve this.
'''
```

```python
year=int(input('Enter year: '))

if year%4==0 :
    if year%100==0:
    
        if year%400==0:
            print('leap year')
        else:
            print('not a leap year')
    else:
        print(' leap year')
else:
    print('not leap year')
```
