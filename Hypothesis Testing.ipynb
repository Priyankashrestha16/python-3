```python
'''
Types of Hypothesis Testing
-------------------------------
1. T-Test
   a. One Sample T-Test
   b. Two Sample T-Test

2. ANOVA( Analysis of Variance) Test  
3. Chi-Square Test 



- Statistics
- Probability

stats -> ( Actual mean - Sample mean) / (sd/ root(total_number _of_data))

- if stats value increases P-value decreses
- if stats value decreses P-value increses




'''
```

```python
# Statistical calculation -> scipy -> scientific python

from scipy import stats
import pandas as pd
```

```python
df = pd.read_csv('Dataset/pizza_sales.csv')
df.head()
```

#### One Sample T-Test

##### Is the average total price of all pizza sales different from $15?

```python
alpha= 0.05
sample = 15
population = df['total_price']
```

```python
# using formula of stats

import numpy as np

actual_mean = population.mean()
sd= population.std()
total_data= np.sqrt(48619)

top_val= actual_mean - sample
bot_val= sd/ total_data

top_val / bot_val
```

```python
# function for One Sample T-Test

# stats.ttest_1samp(population, pomean=sample)

# stats.ttest_1samp(population, popmean=sample)
t_statistics, p_value= stats.ttest_1samp(population, popmean=sample)
```

```python
ho= f'The average total price of all pizza sales equals tp ${sample}.'
ha= f'The average total price of all pizza sales is not equals to ${sample}.'

print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f'Decision: Failes to accept null hypothesis. \n Conclusion: Needs to accpet alternate hypothesis. \n Ha: {ha}')
else:
    print(f'Decision: Failes to reject null hypothesis. \n Conclusion: Needs to accpet null hypothesis. \n Ho: {ho}')
```

#### Two Sample T test

#### Is the average unit price of Classic pizza and chicken pizaas same?

```python
alpha= 0.05
classic_price= df[df['pizza_category']== 'Classic']['unit_price']
chicken_price= df[df['pizza_category']== 'Chicken']['unit_price']
```

```python
classic_price
```

```python
chicken_price
```

```python
# Function for Two Sample T test

# stats.ttest_ind(category1, category2, equal_var= True/False)

t_statistics,p_value=stats.ttest_ind(classic_price, chicken_price, equal_var=False)
p_value
```

```python
ho= f'The average unit price of Classic Pizza and Chicken Pizza are equal.'
ha= f'The average unit price of Classic Pizza and Chicken Pizza are not equal.'

print(f'Ho: {ho}')
print(f'Ha: {ha}')
```

```python
if p_value < alpha:
    print(f'Decision: Failes to accept null hypothesis. \n Conclusion: Needs to accpet alternate hypothesis. \n Ha: {ha}')
else:
    print(f'Decision: Failes to reject null hypothesis. \n Conclusion: Needs to accpet null hypothesis. \n Ho: {ho}')
```

#### ANOVA test

##### Do the average total_price generated by all four pizza categries( Classic, Chicken,  Supreme and Veggie) differ significantly from one another

```python
alpha= 0.05
categories= ['Classic', 'Chicken', 'Supreme','Veggie']
grouped_prices=[]

for category in categories:
    price= df[df['pizza_category']== category]['total_price']
    grouped_prices.append(price)
```

```python
grouped_prices
```

```python
# Function for ANOVA test

t_statistics, p_value= stats.f_oneway(*grouped_prices)
```

```python
pizza_price= df.groupby('pizza_category')['total_price'].mean().rename_axis('Pizza Category').rename(
    'Total Price'
).map('${:,.2f}'.format).to_markdown(numalign='center', stralign='center')
```
