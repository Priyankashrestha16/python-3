```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
employee_name= ['Bob','Arthur','Bill','Duth','Sadie']
employee_salary= [20000,30000,10000,40000,34000]
bonus= [5000, 10000, 2000, 15000, 13500]
```

#### Line Plot

```python
'''
In plot:   
x-axis should have independent variable
y-axis should have dependent variable
'''
```

```python

plt.figure(figsize=(6,4))   # 600 x 400 pixel
plt.plot(employee_name,employee_salary, color= 'green', marker='o', linestyle='-.', label= 'Salary')
                   # hex code can be used for color 
                    # marker and linestyle  have lot more options 
                    # label is used for legend()
plt.plot(employee_name,bonus,color='red', marker='p', label= 'Bonus')
plt.xlabel('Employee Name')
plt.ylabel('Employee Salary / Bonus')
plt.title('Total Salary Distribution by Employees')
plt.grid(alpha=0.8, linestyle='dashed')
plt.xticks(rotation=30)
plt.legend()
plt.savefig('salary_bonus.png')       #saving image of line plot by giving the name for image
plt.show()  # ending the plot
```

#### Bar Plot

```python
plt.figure(figsize=(6,4.5))
plt.bar(employee_name,employee_salary, color=['green','blue','orange','purple','pink'], width=0.5, edgecolor='black')
plt.xlabel('Employee Name')
plt.ylabel('Employee Salary')
plt.title('Total Salary distribution')
plt.xticks(rotation=15)    # rotation of x-label by 15 degree 


# ha( horizontal alignment) -> left/center/right 
# va ( vertical alignment) -> top/center/bottom
for idx,val in enumerate(employee_salary):
    plt.text(idx,val,str(val),ha='center', va='bottom')

plt.show()
```

### Pie chart

```python
df= pd.read_csv('hr_cleaned_data.csv')
df.head()
```

```python
perf_emp= df.groupby('PerformanceScore')['EmpID'].count().reset_index()
perf_emp
```

```python
plt.pie(
    perf_emp['EmpID'],
    labels= perf_emp['PerformanceScore'],
    autopct ='%1.1f%%',
    explode=(0.3,0,0.4,0)         # gets it out of circle (applied on basis of perf_emp)
)
plt.title('% of Employee by Performance')
plt.show()
```

### Donut Chart

```python
plt.pie(
    perf_emp['EmpID'],
    labels= perf_emp['PerformanceScore'],
    autopct ='%1.1f%%',
    wedgeprops={'width':0.4, 'edgecolor':'black'},              # with edge properties  ( accepts dictoneries)
    pctdistance=0.8 ,              # for aligning how percentage are being shown
    explode=(0.3,0,0.4,0)
)
plt.title('% of Employee by Performance')
plt.show()
```

### Histogram

```python
plt.hist(df['Salary'],edgecolor='black')
plt.xlabel('Employee Salary')
plt.ylabel('Frequency / Number od Employees')
plt.show()
```

```python
'''   
Insights of the Histogram table

- Number of Employee who earn low salary is higher then number of employees who earn high salary  -> Right Skew


'''
```

```python
filter_salary= df[df['Salary'] < 85000]    # filtering the data to aside the outliers
```

```python
salary = filter_salary['Salary']
salary_skew= filter_salary['Salary'].skew()
print(salary_skew)    # when skew value is in between 0 - 0.5 it is condidered normal distribution
                       # more than 0.5 is positive skewness and less than 0 is negative skewness
```

```python
mean_salary=salary.mean()
median_salary= salary.median()
```

```python

plt.Figure(figsize=(6,4))
sns.histplot(salary, bins=10, kde= True)  #bins refers to number of bars in figure
                                   # kde -> Kernel Density Estimator

plt.axvline(mean_salary, color='red', label= 'Mean Salary')    #axvline -> axis vertical line
plt.axvline(median_salary, color='green', label=' Median Salary')
plt.legend()
plt.show()
```

```python
# find total employees by each department and gender

dep_gen = df.groupby(['Department','Gender'])['EmpID'].count().reset_index().rename(
    columns ={'EmpID': 'Total Employees'}
)

dep_gen
```

### Sub-Plots

```python
df.columns
```

```python
# Analysis

dep_details= df.groupby('Department').agg(
    Total_Employees= ('EmpID','count'),
    Total_Salary = ('Salary','sum'),
    Total_Absences= ('Absences','sum'),
    Total_Projects= ('SpecialProjectsCount','sum')
).reset_index()

dep_details.columns= dep_details.columns.str.replace('_',' ',regex=False)


dep_details
```

```python
# formatting the values to M/k/B million, thousand, billion

def format_value(val):
    if abs(val)>= 1e9:
        return f'{val/1e9:,.2f}B'
    elif abs(val) >=1e6:
        return f'{val/1e6:,.2f}M'
    elif abs(val)>= 1e3:
        return f'{val/1e3:,.2f}K'
    else:
        return f'{val}'

    
```

```python
fig,axes= plt.subplots(2,2,figsize=(14,10))     #(No_ofrow, no_ofcolumn)

plt.subplots_adjust(hspace=0.5)     # gap between row charts

ax1= axes[0,0]
ax1.bar(dep_details['Department'], dep_details['Total Salary'])
ax1.set_xlabel('Department')
ax1.set_ylabel('Total Salary')
ax1.set_title('Total salary distribution on each department')
ax1.tick_params(axis='x',rotation=15)

for idx,val in enumerate(dep_details['Total Salary']):
    formatted_value= format_value(val)
    ax1.text(idx,val,str(formatted_value),ha='center',va='bottom')

ax2= axes[0,1]
ax2.plot(dep_details['Department'], dep_details['Total Absences'])
ax2.set_xlabel('Department')
ax2.set_ylabel('Total Absences')
ax2.set_title('Total Absences distribution on each department')
ax2.tick_params(axis='x',rotation=15)

for idx,val in enumerate(dep_details['Total Absences']):
    formatted_value= format_value(val)
    ax2.text(idx,val,str(formatted_value),ha='center',va='bottom')

ax3= axes[1,0]
ax3.plot(dep_details['Department'], dep_details['Total Employees'])
ax3.set_xlabel('Department')
ax3.set_ylabel('Total Employees')
ax3.set_title('Total Employees distribution on each department')
ax3.tick_params(axis='x',rotation=15)

ax4= axes[1,1]
ax4.bar(dep_details['Department'], dep_details['Total Projects'])
ax4.set_xlabel('Department')
ax4.set_ylabel('Total Projects')
ax4.set_title('Total Projects distribution on each department')
ax4.tick_params(axis='x',rotation=15)


for idx,val in enumerate(dep_details['Total Projects']):
    ax4.text(idx,val,str(val),ha='center',va='bottom')

plt.show()
```
