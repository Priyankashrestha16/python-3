Array Broadcasting

```python
''' 
Rules of Array Braodcasting
1. all shape and size of arrays must be equal
2, one shape should be one and other must be equal 

or ther should i 1*1 array which would work

'''
```

```python
arr=np.array([
    [1,2,3],
    [4,5,6],
    [6,7,8]
])

data=np.array([ 
    [6,4,3],
    [2,3,1],
    [3,4,2]
])

add= arr+data
print(add)
```
