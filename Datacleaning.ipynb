Data Cleaning

```python
'''  
Process of Data Cleaning
------------------------------
1. Handle Null values
   - Fill Null values
      ->fillna()

      ----------------------------
      Continuous data -> mean /median
      -------------------------------
      To determine which one to use mean/median
      -> Skewness -> use .skew()
          -> 0 to 0.5 (Normal distribution/ Bell curve ) = mean
          -> less than 0 (Left Skewness/ Negative Skewness) or more than0.5 ( Right skewness/ Positive Skewness)=median

       -> calculate mean and median
           if both values are equal or almost equal = mean value
                       or else = median value

        -> fill custom record if any values ar enot correct to fill
       
        ----------------------------
        Categorical/ Discrete data -> mode
        ------------------------------

   - Drop NUll values
       ->dropna()
       -> if data comsists of more than 70% null values then drop column.

2. Remove duplicate records
   - find duplicate records -> duplicated()
   - remove duplicate records -> drop_duplicates()

3. Remove unwanted data
  - remove unwanted characters
    -> str.replace()

  - remove whitespace
   -> str.strip()

  - text transformation
    ->  str.upper() / str.lower() / str.capitalize()

'''
```

```python
'''  
to change data permanently
  inplace - True / flase ( default)

Dropna Arguments
-----------------
1. inplace = True/ Flase
2. axis = 0 ( row) / 1 ( column)
3. how = any ( default) / all   --> any means anyone data in row or column is null then entire row/ column is dropped
                                --> all means if all data in row/ column is null then entire row/ column is dropped 


'''



df.dropna(axis=0,how='any',inplace=True)
```

### Fill Null Values

```python
commission_skew= df['COMMISSION_PCT'].skew()
commission_skew
```

```python
commission_mean= round(df['COMMISSION_PCT'].mean(),2)
commission_median= df['COMMISSION_PCT'].median()

print(f'Commission mean ={ commission_mean}')
print(f'Commission Median= {commission_median}')
```

```python
df['COMMISSION_PCT'].fillna(commission_mean)  # not permanenetly filled
```

```python
df["COMMISSION_PCT"]
```

```python
# permanently filleed
#  df['COMMISSION_PCT'].fillna(commission_mean,inplace= True)
#df['COMMISSION_PCT']=df['COMMISSION_PCT'].fillna(commission_mean,inplace= True) # rreplcing the changes to same column
df.fillna({'COMMISSION_PCT': commission_mean},inplace= True)
```
