```python
'''
Case Scenario 3: Automated Password Compliance Validator
• Core Concept: Loops (while loops with conditional exit control)
• Difficulty: Intermediate
• Scenario: A secure network gateway forces users to choose an alphanumeric passcode that 
fulfills two rigid compliance filters: it must be at least 8 characters long, and it cannot 
contain blank space characters.
• Task: Write a program that uses an indefinite loop to prompt the user to input a password. 
If the input fails to meet either constraint, print the exact reason for rejection and decrement 
a total variable called allowed_attempts (starting at 5). The loop should break immediately 
if a successful passcode is provided, or if total allowed attempts reach 0, print a security 
lockout warning.

'''
```

```python

allowed_attempts=5
while True:
    pw=input('Enter password: ')
    if len(pw)>=8:
        if " " in pw:
            print('Password should not contain blank space.')
            allowed_attempts= allowed_attempts-1
            print(f'Remaining attempts: {allowed_attempts}')
        else:
            print('Valid')
            break
    else:
        print('Password length should be at least 8 characters.')
        allowed_attempts= allowed_attempts-1
        print(f'Remaining attempts: {allowed_attempts}')

    if allowed_attempts==0:
        print('security logout.')
        break

        # allowed_attempts= allowed_attempts-1
        # if allowed_attempts==0:
        #     print('Security lockout warning')
```
