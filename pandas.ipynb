```python
import pandas as pd

#1D -> series
#2D -> dataframe
```

```python
#Series is a one-dimensional labeled array capable of holding any data type (integers, strings, floating point numbers, Python objects, etc.). The axis labels are collectively referred to as the index. A Series is like a fixed-size dict in that you can get and set values by index label.

alphabets=pd.Series(['a','b','c','d','e','f','g','h','i','j'])
alphabets
```

```python
# Dataframe is a two-dimensional labeled data structure with columns of potentially different types. You can think of it like a spreadsheet or SQL table, or a dict of Series objects. It is generally the most commonly used pandas object.

products= pd.DataFrame([
    ['Watch','Navyforce',2018,'Rs.35000'],
    ['Tv','MI',2022,'Rs.45000'],
    ['Mobile','Samsung',2023,'Rs.25000'],
    ['Car','Tata',2026,'Rs 100000']
],columns=['Product Name','Brand','Year','Price'],       #for naming the cloumns
index=['A','B','C','D'])              # for naming the rows we use index variable in pandas 

products
```

```python
products.columns=['Pname','Brand','Year','Price']      # alphabet indexing 
products.index=['D','B','C','E']
products
```

Slicing in pandas

# Loc and ILoc


```python
''' 
Loc -> Location       (( alphabet index))
ILoc -> Integer Location     (( numeric index))

'''
```

```python
alphabets[0:3]
```

```python
products[1:3]
```

```python
products.iloc[1:,0:3]      # --> [row slicing , column slicing]
```

```python
products.loc['C':'E','Brand':]
```

```python
products.iloc[
    [0,1],
    [0,2]
]
```

```python
products.loc[
    ['D','C'],
    ['Year','Price']
]
```
