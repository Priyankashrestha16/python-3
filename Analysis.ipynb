```python
'''  
Group by
Aggregate FUnctions -> (mean,max,min,count,sum)

--------------------------------------------------
Order by -> sort values( ascending-=True/false, by= coulmn_name, ignore_index==True/False (default))
rename -. dictonary{'current_column_name': 'replaced_column_name'}
agg -> for anlaysis of multiple aggregate function columns

'''
```

```python
import pandas as pd 
```

```python
df=pd.read_csv('hr_cleaned_data.csv')
df.head()
```

```python
df.columns
```

```python
# find total employees hired through each Recruitment Source

'''  
syntax of Groupby
----------------------

-> dataframe.groupby(column_name)[column_name].agg_funciton()

'''
```

```python
df['RecruitmentSource'].unique()
```

```python
emp_recruitment=df.groupby('RecruitmentSource')['EmpID'].count()  # counting EmpID for each recruitment source 
```

```python
emp_recruitment
```

```python
emp_recruitment=df.groupby('RecruitmentSource')['EmpID'].count().reset_index()
emp_recruitment     # showing index ---- key value
```

```python
emp_recruitment=df.groupby('RecruitmentSource')['EmpID'].count().reset_index().rename(
    columns={'EmpID':'Total Employees','RecruitmentSource':'Recruitment Source'}
)
emp_recruitment     # nwo renaming the column names
```

```python
# all above analysis can be done at one with sorting like here

emp_recruitment=df.groupby('RecruitmentSource')['EmpID'].count().reset_index().rename(
    columns={'EmpID':'Total Employees','RecruitmentSource':'Recruitment Source'}
).sort_values(ascending=True,by='Total Employees',ignore_index=True)
emp_recruitment
```

```python
emp_recruitment=df.groupby('RecruitmentSource')['EmpID'].count()
format_data=emp_recruitment.to_markdown(numalign='center',stralign='center')

print(format_data)
```

```python
# find total employees hired through each recruitment Source and Gender  also salary_label

hire_emp=df.groupby(['RecruitmentSource','Gender','Salary_Label'])['EmpID'].count().reset_index().rename(
    columns={'EmpID':'Total Employees'}
)
hire_emp
```

```python
# find total employees hired and total salary distribution of employees through each recruitment Source and Gender

# using agg to aggregate multiple columns

hire_emp_salary=df.groupby(['RecruitmentSource','Gender']).agg(
    Total_Employees= ('EmpID','count'),
    Total_salary=('Salary','sum'),
    Maximum_Salary=('Salary','max'),
    Minimum_Salary=('Salary','min')
).reset_index()


hire_emp_salary['Total_salary']= hire_emp_salary['Total_salary'].map('Rs.{:,.2f}'.format)
hire_emp_salary
```

```python
def addCurrency(dataframe, prefix, *args):
    for col in args:
        dataframe[col]=dataframe[col].map(f'{prefix}''{:,.2f}'.format)
```

```python
hire_emp_salary=df.groupby(['RecruitmentSource','Gender']).agg(
    Total_Employees= ('EmpID','count'),
    Total_salary=('Salary','sum'),
    Maximum_Salary=('Salary','max'),
    Minimum_Salary=('Salary','min')
).reset_index()


addCurrency(hire_emp_salary,'Rs.','Total_salary')
addCurrency(hire_emp_salary,'Rs.','Maximum_Salary')
addCurrency(hire_emp_salary,'Rs.','Minimum_Salary')
hire_emp_salary
```

```python
# find total number of employees hired in each year

emp_hired=df.groupby('HireYear')['EmpID'].count().reset_index().rename(
    columns={'EmpID':'Total Employees'}
)

#pct_change()     -> growth per year by percentage

emp_hired['Hire%']=round(emp_hired['Total Employees'].pct_change().fillna(0),2)*100
emp_hired
```

```python
# DateTime -> YID (Year to Date),MID(Month to Date),DID(Day to Date)
```

```python
# month wise total hired employees 

emp_month_hire=df.groupby(['HireMonth','HireMonth_name'])['EmpID'].count().reset_index().rename(
    columns={'EmpID':'Total Employees'}
)[['HireMonth_name','Total Employees']]     # group on the basis of hiremonth also but only selecting two specific columns to display
emp_month_hire
```

```python
## day wise total employee s hired

# Pandas Category -> pd.Categorical(column_name, categories)

days= ['Sunday','Monday','Tuesday','Wednesday','Thursday','Friday','Saturday']
df['HireDay_name']= pd.Categorical(df['HireDay_name'],categories=days)
df['HireDay_name'].unique()
```

```python
emp_month_hire=df.groupby('HireDay_name', observed=False)['EmpID'].count().reset_index().rename(
    columns={'EmpID':'Total Employees'}
)    # group on the basis of hiremonth also but only selecting two specific columns to display
emp_month_hire
```
