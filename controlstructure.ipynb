Match case

final case is represented by '_'

```python
print('1.Breakfast')
print('2.Dessert')
print('3.Dinner')
print('4.Launch')
print('5.Snacks')
print('6. Drinks')

choice= input('Enter your choice: ')

match choice:

    case '1':
        print('Your have ordered breakfast')
        print('1.Tea')
        print('2.Coffee')
        print('3.Milk')
        b_choice=input('Enter breakfast choice: ')
        match b_choice:
            case '1':
                print('You get Tea')
            case '2':
                print('You get coffee')
            case '3':
                print('You get Milk')
            case _:
                print('Invalid order')
    case '2':
        print('You have ordered Dessert.')
        print('1.Icecream')
        print('2.Sweet')
        d_choice=input('Enter dessert choice: ')
        match d_choice:
            case '1':
                print('You get Icecream')
            case '2':
                print('You get Sweet')
            case _:
                print('Invalid order')
    case '3':
        print('You hvae ordered Dinner.')
    case '4':
        print('You hvae ordered Launch.')
    case '5':
        print('You hvae ordered Snacks.')
    case '6':
        print('You hvae ordered Drinks.')
    case _:
        print('You hvae no order.')
```

```python
#Wap related to e commerce shipping problem
#an online store calculates delivery charges dyanamically. Orders with a cart value over $150 receive free shipping. If the cart value is under $150
#look up the destination region "Domestic" shipping costs a flat fee of $10 while "International" shipping region costs $25. However if the customer 
#holds a premium membershiop pass they receiuve a $5 discouint on any calculated shipping fee under $150 but the final shipping cost can never drop below $0

value=int(input('Enter the cart value:'))

if value>=150:
    print('Free Shipping.')
elif value<=150 and value>=0:
    print('1.Domestic')
    print('2.International')
    d_region=input('Enter the region')
    print('1.Premium')
    print('2.Basic')
    c_membership=input('Enter customer membership:')
    
    if d_region=='Domestic' and c_membership=='premium':
        value+=10
        s_fee=value-5
        print(f'The shipping cost:{s_fee}')
    elif d_region=='International' and c_membership=='premium':
        value+=25
        s_fee=value-5
        print(f'The shipping cost:{s_fee}')
    elif d_region=='International':
        value+=25
        print(f'The shipping cost:{value}')
    elif d_region=='Domestic':
        value+=10
        print(f'The shipping cost:{value}')
    else:
        print(f'Destination Region or Membership is incorrect.')
else:
    print('Shipping cost should never drop below zero')
    
    
```
