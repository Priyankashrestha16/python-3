Conditional Filtering

```python
import pandas as pd
```

```python
df= pd.read_csv('Dataset/Employees.csv')
df.head()
```

```python
''' 
Type of data

1. Numeric    (AND condition)
  - Discrete   --> Whole NUmers(positive/Negative)  
  -Continuous   --> whole Numbers/ decimal numbers

2. Categorical   (OR condiion)
  - Nominal    -> category thay can be compared
  - Ordinal    -> category that cannot be compared 


'''
```

Numeric data type

```python
## find details of employees whose salary is in range of 10000 to 18000

emp_salary=df[(df['SALARY']> 10000) & (df['SALARY']<18000)]
emp_salary

# df[emp_salary]
```

```python
'''  
Between

Range-- Inclusive  

'''
```

```python
emp_sal_between= df[df['SALARY'].between(10000,18000)]
emp_sal_between
```

Categorical

```python
# find employees who eorks under job AD_VP/ IT_PROG

emp_job= df[(df['JOB_ID']== 'AD_VP') | (df['JOB_ID']== 'IT_PROG')]
emp_job
```

```python
'''  
ISIN



'''
```

```python
emp_job_in=df[df['JOB_ID'].isin(['AD_VP','IT_PROG'])]
emp_job_in[['FIRST_NAME','EMAIL']]    # specific colum
```

```python
## find details of employees who works under job IT_PROG and whose salary is in range of 4000 to 7000

emp_detail=df[(df['JOB_ID']== 'IT_PROG') & (df['SALARY'].between(4000,7000))]
emp_detail
 
```

```python
Hr=pd.read_csv('Dataset/HR_Dataset Refresh.csv')
Hr.head()
```

##### Write a Pandas statement to filter and isolate all rows where the Department is exactly "IT/IS".

```python
hr_dep= Hr[Hr['Department']=='IT']
hr_dep
```

##### Write a query to find all employees who are in the "Production" department AND have an EmpSatisfaction score less than 3.

```python
hr_and= Hr[(Hr['Department']=='Production') & (Hr['EmpSatisfaction']<3)]
hr_and
```

##### Filter for records where the employee was sourced via "Glassdoor" OR via "Employee Referral".

```python
hr_or= Hr[(Hr['RecruitmentSource']=='Glassdoor') | (Hr['RecruitmentSource']== 'Employee Referral')]
hr_or
```

##### Filter for all employees who have a Salary greater than 100,000 AND have logged more than 10 Absences. Display only their Employee_Name, Position, and ManagerName

```python
hr_range= Hr[(Hr['Salary'] > 100000) & ( Hr['Absences'] > 10)]
hr_range[['Employee_Name','Position','ManagerName']]
```

##### Find all employees who work under the Manager "Kissy Sullivan" and have been late at least once (DaysLateLast30 is greater than 0)
