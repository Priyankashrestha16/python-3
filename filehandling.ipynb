List to CSV file

creat CSV file 

```python
import csv
```

```python
countries=[
    ['Country','Destination','Price'],
    ['Nepal','KTM','20k'],
    ['Nepal','KVT','20k'],
    ['Nepal','BKT','20k'],
    ['Nepal','Lalitpur','20k']
]
```

```python
# open(file_name,operation, new line)

# newline is added in order to fill the gaps between rows

with open ('countries.csv','w', newline='') as file:   #name of the file
    writer= csv.writer(file)   #convert file into CSV format
    writer.writerows(countries)  #storing content in sperate row

 
```

Read CSV file

```python
with open ('countries.csv','r') as file:
    reader= csv.reader(file)
    for row in reader:

        print(row)
```

Dictonary to CSV file

```python
employees=[
    {
        'Name':'BOb',
        'Salary':'30k',
        'Type':'Junior'
    },

    {
        'Name':'Marie',
        'Salary':'20k',
        'Type':'Junior'
    },

    {
        'Name':'John',
        'Salary':'32k',
        'Type':'Junior'
    }
]
```

```python
with open('employees.csv','w',newline='') as file:   # file name
    colums=['Name','Salary','Type']       # cloumns header names  appear at very first row of the file
    writer= csv.DictWriter(file, fieldnames=colums)   # create file and initialize column header
    writer.writeheader() #write column header
    writer.writerows(employees)   #values in rows
```

Read DIctonary file

```python
with open('employees.csv','r') as file:
    reader= csv.DictReader(file)
    for row in reader:
        print(row)
```

Append Dict in csv file

```python
new_employee=[
    {

        'Name':'John Cena',
        'Salary':'100k',
        'Type':'Senior'
    
    }
]
```

```python
with open('employees.csv','a',newline='') as file:   # file name
    colums=['Name','Salary','Type']       #header names
    writer= csv.DictWriter(file, fieldnames=colums)   # create file and initialize header
    writer.writerows(new_employee)
```

Case 4:

7. Dynamic CSV Header Mapper * Task: Create a dictionary representing a student: 
student_data = {"ID": 101, "Name": "Sita", "Grade": "A"}.
• Step 2: Write a script that automatically extracts the keys from this dictionary to use 
as fieldnames for a csv.DictWriter.
• Step 3: Save the data into students.csv.

```python
student_data=[
    {
        'ID':101,
        'Name':'Sita',
        'Grade':'A'

    },
    {
        'ID':102,
        'Name':'gita',
        'Grade':'B'

    },
    {
        'ID':103,
        'Name':'rita',
        'Grade':'A+'

    },
    {
        'ID':104,
        'Name':'nita',
        'Grade':'C'

    }
    

]
```

```python
with open('student.csv','w',newline='') as file:
    column=['ID','Name','Grade']      # it should be the same name as the key in the dictonary
    writer=csv.DictWriter(file,fieldnames=column)
    writer.writeheader()
    writer.writerows(student_data)
```

```python
with open('student.csv','r') as file:
    reader=csv.DictReader(file)
    for rows in reader:
        print(rows)
```

JSON 

```python
import json
```

```python
employees_json=[
    {
        'Name':'BOb',
        'Salary':'30k',
        'Type':'Junior'
    },

    {
        'Name':'Marie',
        'Salary':'20k',
        'Type':'Junior'
    },

    {
        'Name':'John',
        'Salary':'32k',
        'Type':'Junior'
    }
]
```

```python
#dump() ---> store data     (write operation)
#load()  ---> read data      (read operation)
```

Create file

```python
with open('employee.json','w') as file:
    json.dump(employees_json,file, indent=1)   #indent prevents the all data  from being stored on the same line
```

Read File

```python
with open('employee.json','r') as file:
    data=json.load(file) 
    for row in data:
        print(row)
```

Append File

```python
new_employee_json={

        'Name':'John Cena',
        'Salary':'100k',
        'Type':'Senior'
    
    }

data
    
```

```python
with open('employee.json','w') as file:      #JSON doesnot have the concept of append or update directly so we need to perform 'w' operation
    data.append(new_employee_json)
    json.dump(data,file)
```
