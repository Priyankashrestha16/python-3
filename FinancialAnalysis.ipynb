### Loan Status Grid View

```python
# Currrency Converter Function

def currconverter(dataframe, prefix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map(f'{prefix}''{:,.2f}'.format)

# Percentage Converter Function

def pctconverter(dataframe, suffix, *args):
    for col in args:
        dataframe[col] = dataframe[col].map('{:,.2f}'f'{suffix}'.format)
```

```python
MTD_fund= df[df['Issue_Month']== 12]
data=MTD_fund.groupby('loan_status')['loan_amount'].sum().reset_index()
data
```

```python
received_fund= MTD_fund.groupby('loan_status')['total_payment'].sum().reset_index()
```

```python
loan_grid= df.groupby('loan_status').agg(
    Total_Loan_Applicants=('id','count'),
    Total_Funded_Amount=('loan_amount','sum'),
    Total_Amount_Received=('total_payment','sum'),
    Total_Interest_Rate =('int_rate','sum'),
    Total_DTI=('dti','sum')
).reset_index()

# merging analyzed data columns
loan_grid['MTD_Funded_Amount']= data['loan_amount']
loan_grid['MTD_Received_Amount']= received_fund['total_payment']

# replacing unwanted value in columns name
loan_grid.columns= loan_grid.columns.str.replace('_',' ',regex=False)


#usingfunctions

currconverter(loan_grid,'$','Total Funded Amount','Total Amount Received','MTD Funded Amount','MTD Received Amount')
pctconverter(loan_grid,'%','Total Interest Rate','Total DTI')

loan_grid
```

### Analysis

Monthly Trends by Issue Date

```python
month_analysis=df.groupby(['Issue_Month','Issue_Mname']).agg(
    Total_Loan_Applicants= ('id','count'),
    Total_Funded_Amount= ('loan_amount','sum'),
    Total_Received_Amount= ('total_payment','sum')
).reset_index().rename(columns= {'Issue_Mname':'Month Name'})

# replacing unwanted value
month_analysis.columns= month_analysis.columns.str.replace('_',' ',regex=False)

#using function for currency additionin columns
currconverter(month_analysis,'$','Total Funded Amount','Total Received Amount')

# not taking issue_MOnth and only displaying specific columns
month_analysis[['Month Name', 'Total Loan Applicants',
       'Total Funded Amount', 'Total Received Amount']]
```
