```python
import sqlite3
```

```python
conn = sqlite3.connect('ecommerce.db')       #connecting database with python
```

```python
cursor=conn.cursor()     #data store and retrive between python and database through queries
```

Create Table

```python

try:
    cursor.execute('''
        CREATE table customers(
        customer_if integer primary key autoincrement,
        customer_name text,
        customer_phone text,
        customer_address text

    );
    
    ''')
    print('Customer table created.')
except sqlite3.OperationalError as e:
    print(e)
```

```python
try:
    cursor.execute('''
        CREATE table staffs(
        staff_id integer primary key autoincrement,
        staff_name text,
        staff_phone text,
        job_position text,
        job_type text

    );
    
    ''')
    print('Staff table created.')
except sqlite3.OperationalError as e:
    print(e)
```

```python
# CUSTOMER makes ORDER and STAFF manages ORDER
try:
    cursor.execute('''
        create table orders(
            order_id integer primary key autoincrement,
            order_status text,
            order_data date,
            order_quanitty integer,
            order_price real,
            customer_id integer,
            staff_id integer,
            foreign key (customer_if) references customers(customer_id),
            foreign key (staff_id) references staffs(staff_id)
            )

    ''')
    print('Order table created successfully.')
except sqlite3.OperationalError as e:
    print(e)
```

Insert Data

```python
try:
    cursor.execute('''
            insert into customers (customer_name,customer_phone,customer_address)
            values 
            ('Ram','9436523312','Bhaktapur'),
            ('Sita','9247824682','Lalitpur');

        ''')
    conn.commit()              # commit() is used to permanently save the output/results
    print('Customers data inserted Successsfully.')
except sqlite3.OperationalError as e:
    print(e)
```

Update Data

```python
try:
    cursor.execute('''
            update customers set customer_phone='635163413', customer_address='Jhapa' where customer_if=3;

        ''')
    conn.commit()              # commit() is used to permanently save the output/results
    print('Customers data inserted Successsfully.')
except sqlite3.OperationalError as e:
    print(e)
```

Delete data

```python
try:
    cursor.execute('''
            delete from customers where customer_if=2;

        ''')
    conn.commit()              # commit() is used to permanently save the output/results
    print('Customers data inserted Successsfully.')
except sqlite3.OperationalError as e:
    print(e)
```
