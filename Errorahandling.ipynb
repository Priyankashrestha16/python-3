Custom exception

```python
class Pinerror(Exception):
    pass

class Balancenotenougherror(Exception):
    pass

class Balanceerror(Exception):
    pass

```

```python

def atm():
    pinn='1234'
    amountn=60000

    pin=input("Enter the pin number: ")
    amount=int(input("Enter amount: "))

    if pin==pinn:
        if amount<= amountn:
            if amount%1000==0:
                print("Amount withdraw.")
                print(f'Withdrawn amount={amount}.')
                print(f'Remaining Amount={amountn-amount}.') 

            else:
                raise Balanceerror("Error: Amount should be multiple of 1000.")
        else:
            raise Balancenotenougherror("Error:Insufficent amount.")
    else:
        raise Pinerror('Error: Incorrect pin.')
```

```python
pinn='1234'
amountn=60000

try:
    atm()
except Pinerror as pe:
    print(pe)
except Balanceerror as be:
    print(be)
except Balancenotenougherror as bne:
    print(bne)
except ValueError as ve:
    print(f'Error: only number input can be inserted to balance field.')
```
