```python

from selenium import webdriver  # wendriver used to access webbrowser
from selenium.webdriver.common.by import By #used to access website elements/ html tags

```

```python
website='https://merolagani.com/LatestMarket.aspx'
```

access web browser

```python
driver= webdriver.Chrome()
```

```python
driver.get(website)
```

```python
thead=driver.find_element(By.TAG_NAME,'tbody')
```

```python
table_rows= thead.find_elements(By.TAG_NAME,'tr')
```

```python
# for row in table_rows:
#     columns=row.find_elements(By.TAG_NAME,'td')
#     for col in columns:
#         print(col.text)
```

```python
symbol_col=[]
ltp_col=[]
change_col=[]
open_col=[]
high_col=[]
low_col=[]
qty_col=[]
pclose_col=[]
dif_col=[]
```

```python
for row in table_rows:
    columns=row.find_elements(By.TAG_NAME,'td')

    symbol_col.append(columns[0].text)
    ltp_col.append(columns[1].text)
    change_col.append(columns[2].text)
    open_col.append(columns[3].text)
    high_col.append(columns[4].text)
    low_col.append(columns[5].text)
    qty_col.append(columns[6].text)
    pclose_col.append(columns[7].text)
    dif_col.append(columns[8].text)
```

```python
symbol_col
```

```python
ltp_col
```

```python
data ={
    "symbol":symbol_col,
    "ltp":ltp_col,
    "change":change_col,
    "Hight":high_col,
    "low":low_col,
    "open":open_col,
    "qty":qty_col,
    'pclose':pclose_col,
    'diff':dif_col
  }
```

```python
driver.quit()
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data)

df.head(5)
```

```python
df.to_csv('Share.csv',index=False)    #converting to csv format
```

```python
df.to_json('Share.c',indent=4)          # indent reprensents how much gap will be in data
```
